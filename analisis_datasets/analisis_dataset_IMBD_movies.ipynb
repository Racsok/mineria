```python
import pandas as pd
df = pd.read_csv("/content/movie_metadata.csv")
```

```python
def perfilar(df):
    filas = []
    for col in df.columns:
        s = df[col]
        fila = {
            "columna":   col,
            "tipo":      str(s.dtype),
            "distintos": s.nunique(),
            "faltan_%":  round(s.isna().mean() * 100, 1),
            "moda":      s.mode().iloc[0] if not s.mode().empty else None,
        }
        if pd.api.types.is_numeric_dtype(s):
            fila["min"]     = round(s.min(), 2)
            fila["mediana"] = round(s.median(), 2)
            fila["media"]   = round(s.mean(), 2)
            fila["max"]     = round(s.max(), 2)
        filas.append(fila)
    return pd.DataFrame(filas).set_index("columna")
```

## Escriba **tres problemas** que encontró y qué haría con cada uno.
   1. `gross` y `budget` tienen un porcentaje alto de faltantes (17.5% y 9.8%) y además presentan valores atípicos extremos vía IQR (318 atípicos en `budget`, 306 en `gross`); el máximo de `budget` llega a más de 12,000 millones de dólares, una cifra imposible para el presupuesto real de una sola película, lo que sugiere registros en monedas distintas al USD sin convertir. La solución sería primero investigar y corregir esos presupuestos (convertir a USD o marcarlos/eliminarlos si no se puede verificar la moneda), y para los faltantes crear una bandera binaria `gross_faltante` / `budget_faltante` antes de imputar, ya que la ausencia no es aleatoria (ver punto 3).
   2. El dataset tiene **45 filas duplicadas**, lo que puede sesgar las estadísticas descriptivas y sobre-representar ciertas películas en el análisis o en el entrenamiento de un modelo. La solución es eliminarlas con `df.drop_duplicates()`, revisando antes si son duplicados exactos en todas las columnas o solo coinciden en `movie_title` con datos ligeramente distintos (para no borrar películas legítimamente repetidas, como remakes).
   3. La columna `color` es casi constante (solo 2 valores únicos, dominada por "Color") por lo que aporta poca información predictiva por sí sola. Sin embargo, se descubrió que el porcentaje de faltantes en `gross` cambia fuertemente según `color` (33.5% en "Black and White" vs 16.6% en "Color") y según `content_rating` (de 4.2% en PG-13 hasta 100% en varias categorías de TV), es decir, la ausencia de `gross` no es aleatoria (MNAR). La solución no es eliminar `color` sin más, sino conservarla como variable auxiliar/indicador de faltante informativo en vez de imputar `gross` de forma ciega.
   
3. Diga cuál es la **columna objetivo** y su reparto.

La columna objetivo es `imdb_score`. Es una variable continua, por lo que el problema es de **regresión**. Su reparto, según `df["imdb_score"].describe()`:
    Mínimo: 1.60
    Percentil 25%: 5.80
    Mediana (50%): 6.60
    Percentil 75%: 7.20
    Máximo: 9.50
    Media: 6.44 | Desviación estándar: 1.13

A diferencia de una variable con sesgo marcado hacia un extremo, aquí el reparto es bastante simétrico y ligeramente sesgado a la izquierda (la media, 6.44, es apenas menor que la mediana, 6.60), con la mayoría de las películas concentradas entre 5.8 y 7.2 puntos, y colas cortas hacia calificaciones muy bajas o muy altas.

4. Una frase: ¿este archivo aguanta el semestre, sí o no, y por qué?

Sí aguanta el semestre, porque tiene un volumen considerable de datos (5,043 registros, 28 variables con mezcla de tipos: 13 float, 12 object, 3 int), una variable objetivo continua clara (`imdb_score`) para un problema de regresión, y suficientes retos reales de preprocesamiento —faltantes de hasta 17.5% con patrón no aleatorio (MNAR), 45 duplicados, valores atípicos extremos en `budget` y `gross`, alta cardinalidad en columnas de texto y una variable casi constante— como para justificar de sobra las distintas fases de minería de datos.
