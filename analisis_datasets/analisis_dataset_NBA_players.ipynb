```python
import pandas as pd
df = pd.read_csv("/content/all_seasons.csv")
```

```python
def perfilar(df):
    filas = []
    for col in df.columns:
        s = df[col]
        fila = {
            "columna":   col,
            "tipo":      str(s.dtype),
            "distintos": s.nunique(),
            "faltan_%":  round(s.isna().mean() * 100, 1),
            "moda":      s.mode().iloc[0] if not s.mode().empty else None,
        }
        if pd.api.types.is_numeric_dtype(s):
            fila["min"]     = round(s.min(), 2)
            fila["mediana"] = round(s.median(), 2)
            fila["media"]   = round(s.mean(), 2)
            fila["max"]     = round(s.max(), 2)
        filas.append(fila)
    return pd.DataFrame(filas).set_index("columna")
```



## Escriba tres problemas que encontró y qué haría con cada uno.
1) El primer problema que se encontró fue la columna 'unnamed: 0' que representa una fuga de datos y es un indice redundante pues es un indice autogenerado. Se soluciona eliminando directamente la columna con df.drop.
2) Existen valores nullos en 'college' (14.4% de faltantes), existen jugadores que fueron reclutados de la secundaria o internacionalmente. La solución en este caso es pasar de 'NaN' a 'None' o 'International/HighSchool' para no perder esa información categórica.
3) En las variables que destacan al draft como tal, 'draft_year', 'draft_round', 'draft_number'. Se leen como 'object' debido a que existe el texto 'undrafted', esto implide su tratamiento númerico o análisis de correlación directo. La solución en este caso seria crear una bandera binaria is_drafted (1 si si, 0 si no) y convertir 'Undrafted' a 0 en las columnas numéricas para que estas se puedan convertir a int / float.

## Diga cuál es la columna objetivo y su reparto. La columna objetivo es 'pts'. Es una variable continua, su target es de regresión. El reparto se describe mediante distribución continua:

    Mínimo: 0.00 pts
    Mediana: 6.70 pts
    Media: 8.21 pts
    Máximo: 36.10 pts

Este reparto presenta un sesgo positivo hacia la derecha, donde el 50% de los registros anota 6.70 puntos o menos. Una minoría elitista de anotadores eleva la media a 8.21 pts y alcanza un pico en 36.10 pts.

Una frase: ¿este archivo aguanta el semestre, sí o no, y por qué? Sí aguanta perfectamente el semestre porque cuenta con un volumen sólido de datos (12,844 registros), 22 variables con una mezcla rica entre tipos numéricos y categóricos, una variable objetivo clara 'pts' para modelos de regresión, suficientes desafíos de preprocesamiento (nulos, tipados de caracteres y detección de atípicos) para justificar las fases de minería.
